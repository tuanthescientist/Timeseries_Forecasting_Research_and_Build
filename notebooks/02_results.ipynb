# BTC saved results

Read reviewed tables only. Empty tables mean no research results exist. Historical notebook scores and synthetic smoke checks are separate evidence.

```python
import csv
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "results" / "tables").exists():
    ROOT = ROOT.parent
tables = sorted((ROOT / "results" / "tables").glob("*.csv"))
if not tables:
    print("No reviewed BTC research tables yet.")
for path in tables:
    print(path.name)
    with path.open(newline="", encoding="utf-8") as stream:
        for number, row in enumerate(csv.DictReader(stream)):
            if number >= 20:
                break
            print(row)
```
